# Student Performance Analysis & Modeling
## Comprehensive Exploratory Data Analysis & Feature Insights

This notebook conducts an in-depth statistical and exploratory analysis of demographic, socio-economic, and academic factors influencing student exam performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
%matplotlib inline
```

### 1. Data Ingestion & Overview

```python
df = pd.read_csv('../student_data.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

### 2. Statistical Summary & Missing Values

```python
print('Missing Values:\n', df.isnull().sum())
df.describe()
```

### 3. Correlation Analysis Between Academic Domains

```python
score_cols = ['math score', 'reading score', 'writing score']
corr = df[score_cols].corr()
plt.figure(figsize=(6, 5), dpi=150)
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.3f', cbar=True)
plt.title('Inter-Subject Score Correlation Matrix')
plt.show()
```

### 4. Demographic & Socio-Economic Impact Analysis

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5), dpi=150)
sns.boxplot(x='lunch', y='math score', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Lunch Assistance vs Math Score')

sns.boxplot(x='test preparation course', y='math score', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Test Prep Course vs Math Score')

sns.boxplot(x='parental level of education', y='math score', data=df, ax=axes[2], palette='Set2')
axes[2].set_title('Parental Education vs Math Score')
axes[2].tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```
